## **Ejercicio 01: Errores Tipográficos y Valores Inconsistentes: maritalstatus**

### 1.1   Imprime de la columna 'maritalstatus' los valores únicos


```python
print("Valores únicos en maritalstatus:")
print(df['maritalstatus'].unique())
```

### 1.2   Imprime la frecuencia de cada valor único


```python
print("\nFrecuencia de cada valor:")
print(df['maritalstatus'].value_counts(dropna=False)) # dropna=False incluye los valores NaN en el conteo
```

### 1.3   Cambia el valor 'Never-Married' a 'Single'


```python
df['maritalstatus'] = df['maritalstatus'].replace('Never-married', 'Single')
```

### 1.4   Imprime la frecuencia de cada valor único

```python
#Volvemos a imprimir las frecuencias para ver los cambios
print("\nFrecuencia después de la corrección:")
print(df['maritalstatus'].value_counts(dropna=False))
```

## **Ejercicio 02: Errores Tipográficos y Valores Inconsistentes: highestdegree.**

### 2.1 Imprime de la columna 'highestdegrees' los valores únicos.


```python
print("Valores unicos en highestdegree:\n")
print(df['highestdegree'].unique())
```

### 2.2 Imprime la frecuencia de cada valor único



```python
print("Frecuencia de cada valor:\n")
print(df['highestdegree'].value_counts(dropna=False))
```

### 2.3-2.4 Observa que cada valor categórico empieza con 0. Modifica las categorías para que todas estén en minúsculas y ninguna empiece con un número. Por ejemplo, si dice "2. High School", debería ser 'high school. Los cambios se realizarán en la misma columna 'highestdegrees'.



```python
df['highestdegree'] = (
    df['highestdegree']
    .str.lower()                       # pasa todo a minúsculas
    .str.replace(r'^\d+\.\s*', '', regex=True)
    # ^ indica el inicio. \d+ representa digitos del 0-9 en cualqueir cantidad (1,50,1000,9532, etc.).
    #\. indica que habra un punto
    #\s* indica que habra espacios en blanco, quiza mas de 1 (*)
)
```

### 2.5 Imprime los nuevos valores únicos.

```python
print("\nValores únicos después de la limpieza:\n")
print(pd.Series(df['highestdegree'].unique()))
```

## **Ejercicio 03: Alta Cardinalidad - Frecuencias Bajas**



### 3.1 De la columna 'maritalstatus' une las categorías Widowed y 'Single' en una categoría más general llamada 'Single/No Partner'. Crea una nueva columna, 'marital_grouped' donde aparezcan las nuevas categorías de maritalstatus.



```python
df['marital_grouped'] = df['maritalstatus'].replace(
    {'Widowed': 'Single/No Partner',
     'Single': 'Single/No Partner'}
)
```

### 3.2 Imprime los valores únicos de la nueva columna 'marital_grouped'.



```python
print("Valores unicos en marital_grouped:\n")
print(pd.Series(df['marital_grouped'].unique()))
```

### 3.3 Imprime la frecuencia de los valores únicos de la nueva columna 'marital_grouped'

```python
print("Frecuencia en marital_grouped:\n")
print(df['marital_grouped'].value_counts(dropna=False))
```

## **Ejercicio 04: Tipos de Datos**

### 4.1 Imprime el tipo de datos de la columna gender



```python
print("Tipo de dato original de 'gender':", df['gender'].dtype)
```

### 4.2 Imprime la cantidad de bytes de dicha columna


```python
print("Memoria usada por 'gender':", df['gender'].memory_usage(deep=True), "bytes")
```

### 4.3 Si la columna es de tipo object, modifica a category.


```python
if df['gender'].dtype == 'object':
    df['gender'] = df['gender'].astype('category')
```

### 4.4 En caso de haber realizado la modificación del tipo de datos, imprime la cantidad de bytes que ocupa dicha columna con el nuevo tipo de datos

```python
print("Tipo de dato nuevo de 'gender':\n", df['gender'].dtype)
print("Memoria usada ahora:\n", df['gender'].memory_usage(deep=True), "bytes")
```
