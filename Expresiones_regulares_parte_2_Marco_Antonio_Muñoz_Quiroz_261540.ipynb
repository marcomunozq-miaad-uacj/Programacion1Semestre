```python
# ============================================================
# Actividad: Expresiones regulares en Python
# Alumno: Marco Antonio Muñoz Quiroz
# ============================================================

import re

# ------------------------------------------------------------
# Funcion para resaltar cada ejercicio en pantalla
# Ejemplo: Print1. Extraer correos
# ------------------------------------------------------------
def print_resaltado(numero, descripcion):
    print("\n" + "=" * 50)
    print(f"Print{numero}. {descripcion}")
    print("=" * 50)


texto = """
El 10 de agosto de 2025, María Pérez envió un correo electrónico a la dirección maria.perez@email.com solicitando información sobre el producto X-200. Su número de teléfono es +34 612-345-678. Además, visitó la web https://www.ejemplo.com para consultar precios y disponibilidad.
El servidor registró el acceso desde la IP 192.168.1.15 a las 14:23:56, y se detectó un error en el log con código ERR500. En Twitter, el usuario @DataAnalyst comentó: "Los modelos de IA son el futuro #MachineLearning #IA2025". El totaal de la compra fue de $199.99, y el cliente pagó con tarjeta VISA terminada en 1234.
Por otro lado, el DNI de Juan García es 12345678Z y su código postal 28080. Para más información, puede llamar al 911 234 567 o visitar www.informacion.es. Recordamos que las feechas límite para inscripción son 2025-09-30 y 2025-10-15.
""".strip()

# ------------------------------------------------------------
# 1) Correos electronicos
# ------------------------------------------------------------
print_resaltado(1, "Extraer correos electronicos")

pat_email = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"
emails = re.findall(pat_email, texto)

for e in emails:
    print("-", e)


# ------------------------------------------------------------
# 2) URLs con o sin protocolo
# ------------------------------------------------------------
print_resaltado(2, "Extraer URLs")

pat_url = r"\b(?:https?://|www\.)[^\s\"']+\b|\b[A-Za-z0-9.-]+\.[A-Za-z]{2,}(?:/[^\s\"']*)?\b"
urls = list(dict.fromkeys(re.findall(pat_url, texto)))

for u in urls:
    print("-", u)


# ------------------------------------------------------------
# 3) Telefonos internacionales y nacionales
# ------------------------------------------------------------
print_resaltado(3, "Extraer telefonos")

pat_tel_int = r"\+\d{1,3}(?:[\s-]?\d{2,4}){2,4}\b"
pat_tel_nac = r"\b\d{3}(?:[\s-]?\d{3}){2}\b"

telefonos = re.findall(pat_tel_int, texto) + re.findall(pat_tel_nac, texto)

for t in telefonos:
    print("-", t)


# ------------------------------------------------------------
# 4) Fechas formato yyyy-MM-dd
# ------------------------------------------------------------
print_resaltado(4, "Extraer fechas yyyy-MM-dd")

pat_fecha = r"\b\d{4}-\d{2}-\d{2}\b"
fechas = re.findall(pat_fecha, texto)

for f in fechas:
    print("-", f)


# ------------------------------------------------------------
# 5) Codigos postales de 5 digitos
# ------------------------------------------------------------
print_resaltado(5, "Extraer codigos postales")

pat_cp = r"\b\d{5}\b"
cps = re.findall(pat_cp, texto)

for cp in cps:
    print("-", cp)


# ------------------------------------------------------------
# 6) DNI (8 digitos y letra mayuscula)
# ------------------------------------------------------------
print_resaltado(6, "Extraer DNI")

pat_dni = r"\b\d{8}[A-Z]\b"
dnis = re.findall(pat_dni, texto)

for d in dnis:
    print("-", d)


# ------------------------------------------------------------
# 7) Menciones de Twitter
# ------------------------------------------------------------
print_resaltado(7, "Extraer menciones de Twitter")

pat_mencion = r"(?<![\w])@[A-Za-z0-9_]{1,15}\b"
menciones = re.findall(pat_mencion, texto)

for m in menciones:
    print("-", m)


# ------------------------------------------------------------
# 8) Hashtags
# ------------------------------------------------------------
print_resaltado(8, "Extraer hashtags")

pat_hashtag = r"(?<!\w)#[A-Za-z0-9_]+\b"
hashtags = re.findall(pat_hashtag, texto)

for h in hashtags:
    print("-", h)


# ------------------------------------------------------------
# 9) Direcciones IP
# ------------------------------------------------------------
print_resaltado(9, "Extraer IPs")

pat_ip = r"\b(?:\d{1,3}\.){3}\d{1,3}\b"
ips = re.findall(pat_ip, texto)

for ip in ips:
    print("-", ip)


# ------------------------------------------------------------
# 10) Importes con $ y decimales
# ------------------------------------------------------------
print_resaltado(10, "Extraer importes monetarios")

pat_money = r"\$\d+(?:\.\d{2})\b"
montos = re.findall(pat_money, texto)

for m in montos:
    print("-", m)


# ------------------------------------------------------------
# 11) Quitar letras repetidas
# ------------------------------------------------------------
print_resaltado(11, "Eliminar letras repetidas")

texto_limpio = re.sub(r"([A-Za-zÁÉÍÓÚÜÑáéíóúüñ])\1+", r"\1", texto)
print(texto_limpio)
```
