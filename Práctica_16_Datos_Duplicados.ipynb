### Ejercicio 01. Datos Perdidos


```python
# 1.1 Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Semestre1/Programacion/Semana7/Practica16/dirtydata.csv')
print("1.1 Primeras filas del DataFrame:")
print(df.head(), "\n")

# 1.2 Imprime la cantidad de datos perdidos que tiene el conjunto de datos
print("1.2 Valores perdidos por columna:")
print(df.isnull().sum(), "\n")

# 1.3 Calcula la media de la columna Calories
promedio_de_calorias_antes = df["Calories"].mean()
print("1.3 Promedio de 'Calories' ANTES de rellenar:", promedio_de_calorias_antes, "\n")

# 1.4 Aplica imputación por promedio (mean) en la columna Calories
print("1.4 Rellenar valores nulos con la media...\n")
df["Calories"] = df["Calories"].fillna(df["Calories"].mean())

# 1.5 Calcula de nuevo la media de la columna Calories después de que has imputado.
promedio_de_calorias_despues = df["Calories"].mean()
print("1.5 Promedio de 'Calories' DESPUÉS de rellenar:", promedio_de_calorias_despues, "\n")

print("COMENTARIO FINAL: Se observa que el valor de la media no cambia mucho antes y despues de rellenar con la media")
```

### Ejercicio 02. Duplicidad Parcial Una columna: Contar y Filtrar los registros duplicados en Duration

```python
# 2.1 Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
df2 = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Semestre1/Programacion/Semana7/Practica16/dirtydata.csv')
print("2.1 Dimensiones del DataFrame:", df2.shape, "\n")

# 2.2 Cuenta la cantidad de registros duplicados utilizando solo la información de columna Duration e imprime
cantidad_duplicados_duration = df2.duplicated(subset=["Duration"]).sum()
print("2.2 Cantidad de duplicados en 'Duration':", cantidad_duplicados_duration, "\n")

# 2.3 Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.
print("2.3 Conteo de valores únicos en 'Duration':")
print(df2["Duration"].value_counts().sort_index(), "\n")

# 2.4 Filtra los datos y obten los registros sin duplicados
df_sin_duplicados_duration = df2.drop_duplicates(subset=["Duration"], keep="first")
print("2.4 Dimensiones sin duplicados en 'Duration':", df_sin_duplicados_duration.shape,". Ahora son 4 registros y no 32\n")

# Imprimimos nuevamenmte el dataframe para verificar que solo me deja el primer valor que aparezca de cada valor unico en Duration
print("COMENTARIO FINAL: Imprimimos nuevamenmte el dataframe para verificar que solo me deja el primer valor que aparezca de cada valor unico en Duration\n",df_sin_duplicados_duration)
```

### Ejercicio 03 Duplicidad Parcial Dos columnaa: Contar y Filtrar los registros duplicados en Pulse y MaxPulse

```python
# 3.1 Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
df3 = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Semestre1/Programacion/Semana7/Practica16/dirtydata.csv')
print("3.1 Dimensiones del DataFrame:", df3.shape, "\n")

# 3.2 Cuenta la cantidad de registros duplicados utilizando la información de las columnas Pulse y Maxpulse
duplicados = df3.duplicated(subset=["Pulse", "Maxpulse"]).sum()
print("3.2 Cantidad de registros duplicados:", duplicados,"\n")

# 3.3 Utiliza value_counts() e imprime la cantidad de registros que hay por cada valor único.
conteo_valores = df3[["Pulse", "Maxpulse"]].value_counts()
print("3.3 Conteo de combinaciones únicas:\n", conteo_valores,"\n")

# 3.4 Filtra los datos y obten los registros cuando keep = 'last'
datos_filtrados = df2.drop_duplicates(subset=["Pulse", "Maxpulse"], keep="last")
print("3.4 Datos después de filtrar duplicados:\n\n", datos_filtrados.head(),"\n")
print("3.4 Dimensiones nuevas del DataFrame:", datos_filtrados.shape, "\n")

print("COMENTARIO FINAL: el nuevo dataset se redujo de 32 a 29 ya que encontro 3 registros repetidos entre las dos columnas pulse y max pulse")
```

### Ejercicio 04 Duplicidad Total o Exacta: Contar y Filtrar los registros duplicados

```python
# 4.1 Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
df4 = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Semestre1/Programacion/Semana7/Practica16/dirtydata.csv')
print("3.1 Dimensiones del DataFrame:", df4.shape, "\n")

# 4.2 Cuenta la cantidad de registros duplicados exactos (todas las columnas)
duplicados_exactos = df4.duplicated().sum()

# 4.3 Utiliza sum() e imprime la cantidad de registros con duplicidad exacta
print("Cantidad de registros duplicados exactos:", duplicados_exactos,"\n")

# 4.4 Filtra los datos y obten los registros cuando keep = 'last'
datos_filtrados_exactos = df4.drop_duplicates(keep="last")
print("4.4 Datos después de filtrar duplicados exactos:\n", datos_filtrados_exactos.head(),"\n")
print("4.4 Dimensiones nuevas del DataFrame:", datos_filtrados_exactos.shape, "\n")

print("COMENTARIO FINAL: El nuevo dataset se redijo de 32 a 31 registros ya que se elimino el 1 duplicado")
```

### Ejercicio 05: Cálculo de la media con duplicados y sin duplicados

```python
# 5.1 Carga el conjunto de datos dirty.csv y verifica que su carga sea correcta.
df5 = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Semestre1/Programacion/Semana7/Practica16/dirtydata.csv')
print("3.1 Dimensiones del DataFrame:", df5.shape, "\n")

# 5.2 Calcula e imprime la media de las columnas Pulse y MaxPulse
media_pulse = df5["Pulse"].mean()
media_maxpulse = df5["Maxpulse"].mean()
print("Media de Pulse antes de filtrar duplicados:", media_pulse)
print("Media de Maxpulse antes de filtrar duplicados:", media_maxpulse)

# 5.3 Identifica los valores duplicados en las columnas Pulse y MaxPulse (Duplicidad Parcial) y elimina los registro repetidos. Quédate con las últimas ocurrencias
datos_sin_duplicados = df5.drop_duplicates(subset=["Pulse", "Maxpulse"], keep="last")

# 5.4 Utiliza el conjunto anterior sin datos duplicados (recuerda que solo quedaron las últimas ocurrencias de los datos) y calcula la media de las columnas Pulse y MaxPulse
media_pulse_filtrado = datos_sin_duplicados["Pulse"].mean()
media_maxpulse_filtrado = datos_sin_duplicados["Maxpulse"].mean()
print("Media de Pulse después de filtrar duplicados:", media_pulse_filtrado)
print("Media de Maxpulse después de filtrar duplicados:", media_maxpulse_filtrado,"\n")

print("COMENTARIO FINAL: Las medias subieron porque los duplicados que se quitaron tendían a ser más bajos que el promedio.")
```
