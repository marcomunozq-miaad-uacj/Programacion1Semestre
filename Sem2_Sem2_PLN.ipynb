```python
# ============================================================
# modulo de expresiones regulares
# ============================================================
import re
import requests


# ============================================================
# funciones para que la salida se vea ordenada
# ============================================================

def print_title(title):
    # imprime un titulo grande
    line = "=" * 60
    print("\n" + line)
    print(title.upper().center(60))
    print(line)

def show_match(label, m):
    # muestra el resultado solo si existe coincidencia
    if m:
        print(f"{label}: {m.group()}")
    else:
        print(f"{label}: None")

def safe_end(m, default=0):
    # regresa m.end() si existe, si no regresa default
    return m.end() if m else default

def safe_start(m, default=None):
    # regresa m.start() si existe, si no regresa default
    return m.start() if m else default


# ============================================================
# 1) re.match()
# ============================================================

print_title("re.match()")

pattern = r"Cookie"
sequence = "Cookie"

# revisa si la cadena empieza con cookie
if re.match(pattern, sequence):
    print("Match!")
else:
    print("Not a match!")


# ============================================================
# 2) re.search()
# ============================================================

print_title("re.search() basico")

# busca el patron en cualquier parte
show_match("Co.k.e", re.search(r'Co.k.e', 'Cookie'))
show_match("^Eat", re.search(r'^Eat', "Eat cake!"))
show_match("cake$", re.search(r'cake$', "Cake! Let's eat cake"))


# ============================================================
# 3) conjuntos [ ]
# ============================================================

print_title("conjuntos")

show_match("[0-6]", re.search(r'[0-6]', 'Number: 5'))
show_match("[^5]", re.search(r'Number: [^5]', 'Number: 0'))


# ============================================================
# 4) barra invertida
# ============================================================

print_title("barra invertida")

show_match("\\s", re.search(r'Not a\sregular character', 'Not a regular character'))
show_match("caracter normal", re.search(r'Just a \regular character', 'Just a \regular character'))
show_match("literal \\s", re.search(r'Just a \\sregular character', 'Just a \sregular character'))


# ============================================================
# 5) \w y \W
# ============================================================

print_title("w y W")

show_match("\\w", re.search(r'Co\wk\we', 'Cookie'))
show_match("\\W", re.search(r'C\Wke', 'C@ke'))
print("Uppercase W won't match:", re.search(r'Co\Wk\We', 'Cookie'))


# ============================================================
# 6) \s y \S
# ============================================================

print_title("s y S")

show_match("\\s", re.search(r'Eat\scake', 'Eat cake'))
show_match("\\S", re.search(r'cook\Se', "Let's eat cookie"))


# ============================================================
# 7) \d
# ============================================================

print_title("d digitos")

show_match("\\d+", re.search(r'\d+', '100 cookies'))


# ============================================================
# 8) tab y espacios
# ============================================================

print_title("tab y espacios")

# tab real
show_match("tab real", re.search(r'Eat\tcake', 'Eat\tcake'))

# uno o mas espacios
show_match("espacios", re.search(r'Eat\s+cake', 'Eat    cake'))


# ============================================================
# 9) \b
# ============================================================

print_title("b limite de palabra")

show_match("\\b", re.search(r'\b[A-E]ookie', 'Cookie'))


# ============================================================
# 10) cuantificadores
# ============================================================

print_title("cuantificadores")

show_match("Co+kie", re.search(r'Co+kie', 'Cooookie'))
show_match("Ca*o*kie", re.search(r'Ca*o*kie', 'Cookie'))
show_match("Colou?r", re.search(r'Colou?r', 'Color'))
show_match("9 a 10 digitos", re.search(r'\d{9,10}', '0987654321'))


# ============================================================
# 11) grupos ()
# ============================================================

print_title("grupos")

statement = 'Please contact us at: support@datacamp.com'
match = re.search(r'([\w\.-]+)@([\w\.-]+)', statement)

show_match("email", match)

if match:
    print("user:", match.group(1))
    print("host:", match.group(2))
else:
    print("user: None")
    print("host: None")


# ============================================================
# 12) grupos con nombre
# ============================================================

print_title("grupos con nombre")

match = re.search(
    r'(?P<email>(?P<username>[\w\.-]+)@(?P<host>[\w\.-]+))',
    statement
)

if match:
    print("email:", match.group('email'))
    print("user:", match.group('username'))
    print("host:", match.group('host'))
else:
    print("email: None")
    print("user: None")
    print("host: None")


# ============================================================
# 13) greedy vs non greedy
# ============================================================

print_title("greedy vs non greedy")

heading = "<h1>TITLE</h1>"
show_match("greedy", re.match(r'<.*>', heading))
show_match("non greedy", re.match(r'<.*?>', heading))


# ============================================================
# 14) re.compile()
# ============================================================

print_title("re.compile()")

pattern = re.compile(r"cookie")
sequence = "Cake and cookie"

show_match("pattern.search", pattern.search(sequence))
show_match("re.search(pattern)", re.search(pattern, sequence))


# ============================================================
# 15) objeto match
# ============================================================

print_title("objeto match")

print(re.search("cookie", "Cake and cookie"))


# ============================================================
# 16) match vs search
# ============================================================

print_title("match vs search")

print("IceCream:", re.match("C", "IceCream"))
show_match("Cake", re.match("C", "Cake"))


# ============================================================
# 17) findall()
# ============================================================

print_title("findall")

statement = "Please contact us at: support@datacamp.com, xyz@datacamp.com"
emails = re.findall(r'[\w\.-]+@[\w\.-]+', statement)

for e in emails:
    print(e)


# ============================================================
# 18) finditer()
# ============================================================

print_title("finditer")

emails = re.finditer(r'[\w\.-]+@[\w\.-]+', statement)
for e in emails:
    print(e)


# ============================================================
# 19) sub()
# ============================================================

print_title("sub")

statement = "Please contact us at: xyz@datacamp.com"
print(re.sub(r'([\w\.-]+)@([\w\.-]+)', "support@datacamp.com", statement))


# ============================================================
# 20) split()
# ============================================================

print_title("split")

statement = "Please contact us at: xyz@datacamp.com, support@datacamp.com"
pattern = re.compile(r'[:,]')
print(pattern.split(statement))


# ============================================================
# 21) ignorecase
# ============================================================

print_title("ignorecase")

pattern = re.compile("COOKIE", re.IGNORECASE)
match = pattern.search("I am not a cookie monster")

if match:
    print("start:", match.start())
    print("end:", match.end())
    print("span:", match.span())
else:
    print("start: None")
    print("end: None")
    print("span: None")


# ============================================================
# 22) verbose + ignorecase
# ============================================================

print_title("verbose + ignorecase")

statement = "Please contact us at: support@DataCamp.com, xyz@DATACAMP.com"

pattern = re.compile(r"""
[\w\.-]+
@
datacamp.com
""", re.X | re.I)

emails = re.findall(pattern, statement)
for e in emails:
    print("address:", e)


# ============================================================
# 23) project gutenberg
# ============================================================

print_title("project gutenberg")

url = "https://www.gutenberg.org/files/2638/2638-0.txt"

def get_book(url):
    # descarga el texto
    raw = requests.get(url).text

    # busca el inicio real del libro
    m_start = re.search(r"\*\*\* START OF THIS PROJECT GUTENBERG EBOOK .* \*\*\*", raw)
    if not m_start:
        m_start = re.search(r"\*\*\* START OF THE PROJECT GUTENBERG EBOOK .* \*\*\*", raw)

    start = safe_end(m_start, default=0)

    # busca donde empieza II
    m_stop = re.search(r"\nII\b", raw)
    stop = safe_start(m_stop, default=None)

    return raw[start:] if stop is None else raw[start:stop]

def preprocess(text):
    # cambia simbolos por espacios y pasa a minusculas
    return re.sub(r'[^A-Za-z0-9.]+' , ' ', text).lower()

book = get_book(url)
processed = preprocess(book)

print("the count:", len(re.findall(r'the', processed)))

# corrige i por I
processed = re.sub(r'\si\s', " I ", processed)

# cuenta comillas especiales
print("special quotes:", len(re.findall(r'\”', book)))


# ============================================================
# tabla resumen
# ============================================================

# .    cualquier caracter excepto salto de linea
# ^    inicio de cadena
# \A   solo inicio de cadena
# $    final de cadena
# \Z   solo final de cadena
# [ ]  conjunto de caracteres
# \    escape o literal
# \w   letras numeros o _
# \W   contrario de \w
# \s   espacio en blanco
# \S   no espacio
# \d   digito
# \D   no digito
# \t   tabulador
# \n   nueva linea
# \r   retorno de carro
# \b   limite de palabra
# +    una o mas veces
# *    cero o mas veces
# ?    cero o una vez / no codicioso
# { }  numero exacto o rango
# ( )  grupo
# < >  grupo con nombre
```
